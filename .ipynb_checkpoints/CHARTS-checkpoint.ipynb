```python
import os
import pandas as pd
from pymongo import MongoClient
from dotenv import load_dotenv

# Cargamos el .env que ya tenés configurado en la raíz
load_dotenv()

# Construcción de la URI
uri = f"mongodb+srv://{os.getenv('MONGO_USER')}:{os.getenv('MONGO_PASS')}@{os.getenv('MONGO_CLUSTER')}/?appName={os.getenv('MONGO_APP_NAME')}"

try:
    client = MongoClient(uri)
    db = client['Opciones']
    coleccion = db['Data']
    
    # Traemos toda la data de la nube a un DataFrame
    df = pd.DataFrame(list(coleccion.find({})))
    
    if not df.empty:
        df = df.drop(columns=['_id']) # Limpiamos el ID de Mongo
        print(f"✅ Conexión exitosa. Se cargaron {len(df)} registros.")
    else:
        print("⚠️ La base de datos en la nube está vacía.")

except Exception as e:
    print(f"❌ Error: {e}")

# Visualizar los datos en el navegador
df
```

```python
import matplotlib.pyplot as plt

# 1. Configuración inicial
Symbol = "GFGV8230FE"
df_plot = df[df['Symbol'] == Symbol].copy()
df_plot['Date'] = df_plot['Transact Time'].dt.date
dias_disponibles = sorted(df_plot['Date'].unique())

# 2. Loop por cada día detectado
for dia in dias_disponibles:
    df_dia = df_plot[df_plot['Date'] == dia].sort_values('Transact Time')
    
    # Crear la figura y el primer eje (Precio)
    fig, ax1 = plt.subplots(figsize=(14, 6))
    
    # --- Graficar PRECIO (Eje Izquierdo) ---
    color_precio = 'tab:blue'
    ax1.set_xlabel('Hora de Ejecución')
    ax1.set_ylabel('Precio ($)', color=color_precio, fontsize=12, fontweight='bold')
    ax1.plot(df_dia['Transact Time'], df_dia['Executed Value'], color=color_precio, marker='o', label='Precio')
    ax1.tick_params(axis='y', labelcolor=color_precio)
    ax1.grid(True, linestyle='--', alpha=0.5)

    # --- Graficar VOLUMEN (Eje Derecho) ---
    # twinx() crea el segundo eje que comparte el mismo eje X
    ax2 = ax1.twinx() 
    color_size = 'tab:red'
    ax2.set_ylabel('Tamaño (Size)', color=color_size, fontsize=12, fontweight='bold')
    
    # Usamos barras para el volumen para que se distinga bien del precio
    ax2.bar(df_dia['Transact Time'], df_dia['Executed Size'], color=color_size, alpha=0.3, width=0.001, label='Size')
    ax2.tick_params(axis='y', labelcolor=color_size)

    # Título y ajustes finales
    plt.title(f'Evolución {Symbol} | Precio vs Volumen | Fecha: {dia}', fontsize=14)
    fig.tight_layout()
    plt.xticks(rotation=45)
    
    plt.show()
```
