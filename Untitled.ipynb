```python
import pandas as pd
from pymongo import MongoClient

# 1. Conexión y extracción
client = MongoClient("mongodb://localhost:27017/")
db = client["Opciones"]
collection = db["Data"]

# Traemos los documentos (podes filtrar por fecha si tenés muchos)
cursor = collection.find()

# 2. Procesamiento de la estructura anidada
rows = []

for doc in cursor:
    ts = doc.get('timestamp')
    ts_str = doc.get('timestamp_str')
    data = doc.get('data', {})
    
    # Iteramos sobre cada ticker dentro del objeto 'data'
    for ticker, info in data.items():
        # Creamos una fila plana
        row = {
            "timestamp": ts,
            "hora": ts_str,
            "ticker": ticker,
            "bid": info.get('bid'),
            "bid_size": info.get('bid_size'),
            "offer": info.get('offer'),
            "offer_size": info.get('offer_size'),
            "last": info.get('last'),
            "last_size": info.get('last_size')
        }
        rows.append(row)

# 3. Creación del DataFrame
df = pd.DataFrame(rows)

# Convertimos timestamp a objeto datetime real de Pandas si es necesario
df['timestamp'] = pd.to_datetime(df['timestamp'])

# 4. Mostrar resultado
print(f"Total de registros procesados: {len(df)}")
df
```
